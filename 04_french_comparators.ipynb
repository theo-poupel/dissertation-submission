# Attribution baseline: Paris against Lyon and Marseille

PM2.5 declined across all three selected cities, so a fall in Paris alone does not identify a local policy effect. This notebook provides the raw descriptive baseline for the French comparison: station selection, data checks, per-city trends and an untreated-comparator benchmark.

The concentrations in this notebook are not meteorologically normalised. Weather effects are not guaranteed to cancel across cities, so the raw differentials are treated as descriptive benchmarks rather than policy estimates. The later attribution analysis uses meteorologically normalised series.

The study window is 2013-2025. Valid negative readings are retained, and the attribution and mechanism station roles are defined in section 5.

## 1. Setup, the Geod'Air reader, and explicit date parsing

Geod'Air exports switched from dash to slash date format partway through the project, so parsing is explicit per format rather than inferred: inferred parsing chooses per element and can silently transpose day and month. Every load asserts that all timestamps parsed and fall inside the window.

```python
import io
import os

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from linearmodels.panel import PanelOLS

def load_geodair(path):
    with open(path, encoding='utf-8-sig') as f:
        raw = f.read()
    clean = []
    for ln in raw.splitlines():
        if not ln:
            continue
        if ln.startswith('"') and ln.endswith('"'):
            ln = ln[1:-1]
        clean.append(ln.replace('""', '"'))
    df = pd.read_csv(io.StringIO("\n".join(clean)), sep=';', quotechar='"', engine='python')
    df.columns = [c.strip().strip('"') for c in df.columns]
    return df

def parse_geodair_dates(s):
    s = s.astype(str)
    out = pd.Series(pd.NaT, index=s.index, dtype="datetime64[ns]")
    slash = s.str.contains("/", na=False)
    out[slash]  = pd.to_datetime(s[slash],  format="%Y/%m/%d %H:%M:%S", errors="coerce")
    out[~slash] = pd.to_datetime(s[~slash], format="%Y-%m-%d %H:%M:%S", errors="coerce")
    assert out.notna().all(), f"{out.isna().sum()} timestamps failed to parse"
    return out

def load_hourly(folder, city_token, y0=2013, y1=2025):
    """Load valid hourly observations for one city."""
    files = [os.path.join(folder, f"pm2.5_{city_token}_{year}_hourly.csv")
             for year in range(y0, y1 + 1)]
    missing = [path for path in files if not os.path.exists(path)]
    assert not missing, f"{city_token}: missing hourly files: {missing}"
    d = pd.concat([load_geodair(f) for f in files], ignore_index=True)
    d = d[d["validité"] == 1].copy()
    d["valeur brute"] = pd.to_numeric(d["valeur brute"], errors="coerce")
    d = d.dropna(subset=["valeur brute"])
    d["datetime"] = parse_geodair_dates(d["Date de début"])
    assert d["datetime"].dt.year.between(y0, y1).all(), f"{city_token}: timestamp outside window"
    print(f"{city_token}: {len(files)} files, {len(d)} rows, "
          f"{(d['valeur brute'] < 0).sum()} negatives retained")
    return d
```

## 2. Load the three cities

I apply the same loading and cleaning rules to Paris, Lyon and Marseille. The output reports the file count, retained observations, negative valid readings and station coverage for each city.

```python
paris     = load_hourly("Donnee_pm2.5_paris/Donnees_hourly_yearly", "paris")
lyon      = load_hourly("Donnee_pm2.5_lyon", "lyon")
marseille = load_hourly("Donnee_pm2.5_marseille", "marseille")

for name, d in [("PARIS", paris), ("LYON", lyon), ("MARSEILLE", marseille)]:
    cov = d.groupby("code site")["datetime"].agg(["min", "max"])
    cov["first_year"] = cov["min"].dt.year
    cov["last_year"]  = cov["max"].dt.year
    print(f"\n=== {name} ===")
    print(cov[["first_year", "last_year"]].to_string())
```

## 3. Acceptance: self-computed annual means against Geod'Air official

For every station-year where Geod'Air provides an official annual value, I compare it with the mean calculated from the hourly observations. I use 0.10 µg/m³ as a diagnostic tolerance. This is not a Geod'Air rule: it is small relative to the annual concentrations, allows minor rounding or field-convention differences, and would still flag a meaningful mismatch. Years without an official annual value remain visible as missing comparisons and are not counted as passing the check.

The calculation retains valid negative readings, consistent with the analysis rule. Section 10 shows why dropping them produces an upward bias at stations with concentrations near the detection limit.

```python
def acceptance(hourly, yearly_path, stations, label, tol=0.10):
    yr = load_geodair(yearly_path)
    yr = yr[yr["validité"] == 1].copy()
    yr["valeur brute"] = pd.to_numeric(yr["valeur brute"], errors="coerce")
    yr["year"] = yr["Date de début"].astype(str).str[:4].astype(int)
    h = hourly.copy()
    h["year"] = h["datetime"].dt.year
    print(f"\n=== {label} ===")
    worst = 0.0
    for st in stations:
        off  = yr[yr["code site"] == st].set_index("year")["valeur brute"]
        ours = h[h["code site"] == st].groupby("year")["valeur brute"].mean()
        c = pd.DataFrame({"ours": ours, "official": off})
        c["diff"] = c["ours"] - c["official"]
        matched = c["diff"].dropna()
        md_ = matched.abs().max()
        if pd.notna(md_):
            worst = max(worst, md_)
        flag = "" if (pd.isna(md_) or md_ <= tol) else "   *** BREACH"
        missing = int(c["official"].isna().sum())
        print(f"\n{st}: max |diff| = {md_:.3f}{flag} "
              f"({len(matched)} matched years, {missing} without official value)")
        print(c.round(3).to_string())
    print(f"\n{label} worst cell: {worst:.3f}")

P = "Donnee_pm2.5_paris/Donnees_hourly_yearly/pm2.5_paris_2013_2025_yearly.csv"
acceptance(paris, P, ["FR04329","FR04143","FR04055","FR04004","FR04031","FR04131"], "PARIS")
acceptance(lyon,  "Donnee_pm2.5_lyon/pm2.5_lyon_2013_2025_yearly.csv",
           ["FR20062","FR20017","FR20046"], "LYON")
acceptance(marseille, "Donnee_pm2.5_marseille/pm2.5_marseille_2013_2025_yearly.csv",
           ["FR03043","FR03014","FR03006"], "MARSEILLE")
```

## 4. The station census

The attribution comparison should use the same type of monitoring site in every city. I therefore select urban-background stations in each continuous urban core, rather than using every station inside a broad administrative boundary.

A metropolitan-boundary rule would include separate centres such as Aix, Salon and Marignane in the Marseille sample, while municipal boundaries are not comparable across the three cities. The common station type and area fields provide a more consistent basis for selection.

I use the station type, area and municipality fields in the EEA reference metadata. The resulting rule is: include urban-background stations in the main urban core; exclude suburban and rural sites and distinct secondary centres such as Villefranche, Aix, Arles, Salon and Marignane. Traffic stations are reserved for the mechanism analysis.

```python
meta = pd.read_csv("data/meta/stations.csv", low_memory=False)
fr_pm = meta[(meta["Country"] == "France") & (meta["Air Pollutant"] == "PM2.5")]
cols = ["Air Quality Station EoI Code", "Air Quality Station Name",
        "Air Quality Station Type", "Air Quality Station Area", "Municipality"]

for city, pats in [("LYON", ["LYON","VILLEURBANNE","VAULX-EN-VELIN","BRON","VENISSIEUX","CALUIRE"]),
                   ("MARSEILLE", ["MARSEILLE","AIX-EN-PROVENCE"])]:
    m = fr_pm["Municipality"].str.upper().str.contains("|".join(pats), na=False)
    print(f"\n=== {city} candidates ===")
    print(fr_pm[m][cols].drop_duplicates().sort_values("Air Quality Station Type").to_string(index=False))

# stations returned by the department export that the rule excludes
excluded = ["FR20013","FR20037","FR20048","FR20061","FR20077",
            "FR02022","FR02041","FR02043","FR03029","FR03021"]
print("\n=== excluded by the rule ===")
print(meta[meta["Air Quality Station EoI Code"].isin(excluded)][cols]
      .drop_duplicates().sort_values("Air Quality Station EoI Code").to_string(index=False))

# Measurement records used to assess the Paris station handover in section 10e.
instrument_cols = ["Air Quality Station EoI Code", "Sampling Point Id",
                   "Measurement Method", "Measurement Equipment",
                   "Operational Activity Begin", "Operational Activity End"]
paris_splice_meta = (fr_pm[fr_pm["Air Quality Station EoI Code"].isin(["FR04143", "FR04055"])]
                     [instrument_cols].drop_duplicates()
                     .sort_values(["Air Quality Station EoI Code",
                                   "Operational Activity Begin", "Sampling Point Id"]))
print("\n=== Paris splice measurement records ===")
print(paris_splice_meta.to_string(index=False))
```

## 5. Station roles used in the analysis

The attribution tier contains urban-background stations only. Paris FR04329 and Marseille FR03006 are traffic stations and are kept for the mechanism analysis. FR04031 has no role in the specified analysis. FR20013 is retained only as an archived descriptive station because several road and policy changes overlap at that site.

Notebook 03 deliberately used all Paris stations to examine network composition. This notebook instead uses the four urban-background Paris stations listed below. The two samples answer different questions and are not compared directly.

```python
attribution_stations = {
    "Paris":     ["FR04143", "FR04055", "FR04004", "FR04131"],   # urban background only
    "Lyon":      ["FR20062", "FR20017", "FR20046"],
    "Marseille": ["FR03043", "FR03014"],
}
mechanism_stations = {
    "Paris":     ["FR04329"],   # peripherique, traffic
    "Marseille": ["FR03006"],   # Rabatau, traffic
}
# FR04031: no analytic role. FR20013: archived, descriptive appendix only.

def tag(d, city, codes):
    out = d[d["code site"].isin(codes)].copy()
    out["city"] = city
    return out

attribution = pd.concat([tag(paris, "Paris", attribution_stations["Paris"]),
                         tag(lyon, "Lyon", attribution_stations["Lyon"]),
                         tag(marseille, "Marseille", attribution_stations["Marseille"])],
                        ignore_index=True)
mechanism = pd.concat([tag(paris, "Paris", mechanism_stations["Paris"]),
                       tag(marseille, "Marseille", mechanism_stations["Marseille"])],
                      ignore_index=True)

print("Attribution tier (urban background):")
print(attribution.groupby(["city", "code site"]).size())
print("\nMechanism tier (traffic, held separately):")
print(mechanism.groupby(["city", "code site"]).size())
```

## 6. Monthly means per station under the capture rule

I aggregate the hourly observations to station-month means and retain a month when at least 75% of its expected hours are present. The EEA convention is annual, so applying the same percentage to each month is a project choice. It keeps thin months from driving the panel while applying one transparent rule to every station. Monthly means must be positive before taking logarithms.

The panel is unbalanced. Lyon and Marseille each contain at least one station spanning the study window. In the Paris background tier, FR04143 ends in August 2019 and FR04055 first clears the capture rule in December 2019. Section 10 documents this splice and the absence of direct overlap. Station fixed effects allow different average levels, but they cannot create information during the gap or directly calibrate the two stations.

```python
def to_panel(df):
    d = df.copy()
    d["ym"] = d["datetime"].dt.to_period("M")
    g = (d.groupby(["city", "code site", "ym"])["valeur brute"]
           .agg(["mean", "count"]).reset_index())
    g["expected"] = g["ym"].apply(lambda p: p.days_in_month * 24)
    g["capture"]  = g["count"] / g["expected"]
    p = g[g["capture"] >= 0.75].copy()
    p = p.rename(columns={"mean": "pm25", "code site": "station"})
    assert (p["pm25"] > 0).all(), "non-positive station-month mean"
    p["date"] = p["ym"].dt.to_timestamp()
    p["log_pm25"] = np.log(p["pm25"])
    p["month_of_year"] = p["date"].dt.month
    p["t"] = (p["date"].dt.year - 2013) + (p["date"].dt.month - 1) / 12
    return p

panel = to_panel(attribution)
print("Station-months per city:")
print(panel.groupby(["city", "station"]).size())
print("\nTotal:", len(panel), "station-months")
```

## 7. Per-city trends, raw

I estimate a separate station fixed-effects trend for each city, including month-of-year controls. Station fixed effects account for stable level differences between monitoring sites.

The standard errors use the Driscoll-Kraay estimator with a Bartlett kernel. This allows the station residuals to be correlated across sites and over time, both of which are plausible for monthly pollution data. I set the bandwidth to 18 months so that the correction extends beyond one seasonal cycle. The papers motivate the estimator, but 18 months is a project choice rather than a value taken from them.

In the raw attribution sample, the estimated Paris decline is slower than the Lyon and Marseille declines. This is a descriptive result before meteorological normalisation and should not be attributed to policy.

The Paris interval is wider than the comparator intervals. Its staggered station coverage is an important reason for the lower precision, although differing variability between cities can also contribute.

These estimates are used only as a raw baseline. Section 9 compares them with the Lyon-Marseille differential before either comparator's private-car restriction.

```python
def city_trend(sub, label):
    n_stations = sub["station"].nunique()
    assert n_stations >= 2, f"{label}: fewer than two attribution stations"
    indexed = sub.set_index(["station", "date"])
    model = PanelOLS.from_formula(
        "log_pm25 ~ 1 + t + C(month_of_year) + EntityEffects",
        data=indexed, drop_absorbed=True
    ).fit(cov_type="kernel", kernel="bartlett", bandwidth=18)
    beta, se = model.params["t"], model.std_errors["t"]
    ci = model.conf_int().loc["t"]
    trend = (np.exp(beta) - 1) * 100
    lower = (np.exp(ci.iloc[0]) - 1) * 100
    upper = (np.exp(ci.iloc[1]) - 1) * 100
    print(f"{label:22s}: {trend:+.2f} %/yr   "
          f"95% CI [{lower:+.2f}, {upper:+.2f}]   ({n_stations} stations)")
    return {"trend": trend, "beta": beta, "se": se, "model": model}

city_results = {
    city: city_trend(panel[panel["city"] == city], city)
    for city in ["Paris", "Lyon", "Marseille"]
}
trends = {city: result["trend"] for city, result in city_results.items()}
print("\nRaw differentials (Paris minus comparator), baseline only, weather NOT normalised:")
for c in ["Lyon", "Marseille"]:
    print(f"  Paris - {c}: {trends['Paris'] - trends[c]:+.2f} %/yr")
```

## 8. Pooled panel with city-by-time interactions

Overlapping confidence intervals for the separate city trends are not a direct test of their difference. I therefore fit one panel with city-specific time slopes and report the interaction coefficients and their intervals.

The specification contains a common time trend, Lyon-by-time and Marseille-by-time interactions, month-of-year effects and station fixed effects. Paris is the reference city. I multiply each interaction coefficient by 100 and report it as a log-point difference per year. At these magnitudes this is approximately a percentage-point difference in annual trend. A negative coefficient means that the comparator's log trend is steeper than Paris's. I use the same Driscoll-Kraay covariance settings as in section 7 so the separate and pooled results are comparable.

These raw interactions are descriptive. Wide intervals would indicate that differential meteorology and the unbalanced panel leave the unnormalised comparison too imprecise for attribution.

The later weather-normalised analysis resamples calendar years for its main uncertainty calculation. The Driscoll-Kraay intervals here are an analytic check for this raw descriptive baseline.

```python
def fit_city_interactions(data):
    prepared = data.copy()
    prepared["city"] = pd.Categorical(
        prepared["city"], categories=["Paris", "Lyon", "Marseille"]
    )
    indexed = prepared.set_index(["station", "date"])
    return PanelOLS.from_formula(
        "log_pm25 ~ 1 + t + t:C(city) + C(month_of_year) + EntityEffects",
        data=indexed, drop_absorbed=True
    ).fit(cov_type="kernel", kernel="bartlett", bandwidth=18)

def print_city_interactions(model, label):
    print(f"\n{label}")
    print("Log-slope differences, comparator minus Paris:")
    for name in model.params.index:
        if not name.startswith("t:C(city)"):
            continue
        beta = model.params[name]
        ci = model.conf_int().loc[name]
        difference = 100 * beta
        lower = 100 * ci.iloc[0]
        upper = 100 * ci.iloc[1]
        city = name.rsplit("[", 1)[-1].rstrip("]").removeprefix("T.")
        conclusion = "excludes zero" if lower > 0 or upper < 0 else "includes zero"
        print(f"  {city:10s}: {difference:+.2f} log points/yr  "
              f"95% CI [{lower:+.2f}, {upper:+.2f}]  ({conclusion})")

pooled = fit_city_interactions(panel)

print(pooled.summary.tables[1])

print_city_interactions(pooled, "Full 2013-2025 window")
```

## 9. Untreated-comparator benchmark and matched-window comparison

Lyon's private-car restriction began on 1 September 2022, and Marseille's first Crit'Air restriction covering private cars began on the same date. I therefore end the joint pre-restriction comparison in August 2022. The Lyon-Marseille differential over this window is a descriptive indication of the city-to-city variation that can arise before these restrictions.

The untreated window is therefore January 2013 through August 2022. Lyon's earlier freight restrictions are not treated as the private-car onset. This whole-window trend comparison is a descriptive benchmark rather than a formal false-onset placebo.

For scale, I recompute the Paris differentials on the same dates. Paris is not untreated during the whole window: the intra-muros Crit'Air 5 restriction began in July 2017 and Crit'Air 4 in July 2019. The Paris comparisons are therefore not placebo tests.

The pooled matched-window model reports direct intervals for the comparator-minus-Paris log-slope differences. These results remain descriptive because the series are not meteorologically normalised and Paris is already treated for part of the window. A separate false-onset analysis is carried out later using the weather-normalised data.

```python
UNTREATED_END = "2022-08-31"
w = panel[panel["date"] <= UNTREATED_END]

print(f"Untreated window: {w['date'].min().date()} to {w['date'].max().date()}\n")
window_results = {
    city: city_trend(w[w["city"] == city], f"{city} (matched window)")
    for city in ["Paris", "Lyon", "Marseille"]
}
wt = {city: result["trend"] for city, result in window_results.items()}

comparator_gap = wt["Lyon"] - wt["Marseille"]
print(f"\nUNTREATED COMPARATOR DIFFERENCE, Lyon minus Marseille: "
      f"{comparator_gap:+.2f} %/yr")
print("Matched-window Paris differentials, same window:")
for c in ["Lyon", "Marseille"]:
    print(f"  Paris - {c}: {wt['Paris'] - wt[c]:+.2f} %/yr")
print(f"\nRatio of |comparator difference| to the larger |Paris differential|: "
      f"{abs(comparator_gap) / max(abs(wt['Paris']-wt['Lyon']), abs(wt['Paris']-wt['Marseille'])):.1f}x")

matched_pooled = fit_city_interactions(w)
print_city_interactions(matched_pooled, "Direct matched-window comparison")
```

### Figure. Monthly PM2.5 by city, log scale, with fitted trends

The points are monthly geometric means across the capture-clearing attribution stations in each city. This uses the same station-month panel and log outcome as the regressions. The lines show the time component from each city's fixed-effects model, anchored at that city's average log concentration; month-of-year effects are omitted from the plotted lines for readability.

The Paris lines mark the intra-muros Crit'Air 5 and Crit'Air 4 restrictions in July 2017 and July 2019. Lyon and Marseille share a legal private-car restriction onset in September 2022, so one combined line marks that date. The later metropolitan Paris schedule applies to the boulevard périphérique mechanism station and is not used for these background stations.

The month-to-month variation includes meteorology and other common shocks. This is why the plotted raw trends are not interpreted as policy effects.

```python
monthly = (panel.groupby(["city", "date"], observed=True)
           .agg(log_pm25=("log_pm25", "mean"),
                t=("t", "first"))
           .reset_index())

fig, ax = plt.subplots(figsize=(13, 6))
colors = {"Paris": "tab:blue", "Lyon": "tab:orange", "Marseille": "tab:green"}
for city in ["Paris", "Lyon", "Marseille"]:
    s = monthly[monthly["city"] == city].sort_values("date")
    ax.plot(s["date"], s["log_pm25"], ".", color=colors[city],
            alpha=0.35, markersize=4)
    centre_t = s["t"].mean()
    anchor_log = s["log_pm25"].mean()
    trend_line = anchor_log + city_results[city]["beta"] * (s["t"] - centre_t)
    ax.plot(s["date"], trend_line, "-", color=colors[city], linewidth=2,
            label=f"{city}: {city_results[city]['trend']:+.1f} %/yr")

# Ville de Paris intra muros clock (governs the plotted attribution-tier stations)
ax.axvline(pd.Timestamp("2017-07-01"), color="tab:blue", linestyle=":",  alpha=0.6,
           label="Paris Crit'Air 5 (2017-07)")
ax.axvline(pd.Timestamp("2019-07-01"), color="tab:blue", linestyle="--", alpha=0.6,
           label="Paris Crit'Air 4 (2019-07)")
ax.axvline(pd.Timestamp("2022-09-01"), color="tab:purple", linestyle=":", alpha=0.7,
           label="Lyon and Marseille legal onsets (2022-09)")

ax.set_xlabel("Year")
ax.set_ylabel("Log monthly geometric mean PM2.5")
ax.set_title("Monthly PM2.5 by city, attribution stations (2013-2025)")
ax.legend(fontsize=8, loc="upper right")
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### 10d. The Paris background splice

No Paris attribution station spans the complete window. FR04143 ends during 2019 and FR04055 begins later that year, so I check both the hourly boundary and the monthly panel boundary.

The final FR04143 observation and first FR04055 observation are separated by 7 days and 11 hours, with no simultaneous valid hours. A direct cross-calibration is therefore unavailable.

At monthly resolution, neither station clears the 75% rule from September through November 2019. FR04055 first clears it in December. I apply the same threshold at the handover rather than making an exception.

The output prints the hourly endpoints, monthly capture rates and the number of overlapping hours and capture-clearing months.

```python
b143 = paris[paris["code site"] == "FR04143"]["datetime"]
b055 = paris[paris["code site"] == "FR04055"]["datetime"]
print("FR04143 last valid hour: ", b143.max())
print("FR04055 first valid hour:", b055.min())
print("instrument gap:", (b055.min() - b143.max()))

print("\n2019 monthly hours and capture:")
for st in ["FR04143", "FR04055"]:
    d = paris[(paris["code site"] == st) & (paris["datetime"].dt.year == 2019)]
    m = d.groupby(d["datetime"].dt.month).size()
    print(f"  {st}:")
    for mth in range(1, 13):
        hrs = int(m.get(mth, 0))
        exp = pd.Period(f"2019-{mth:02d}", freq="M").days_in_month * 24
        cap = hrs / exp
        mark = "keep" if cap >= 0.75 else ("sub-threshold" if hrs > 0 else "empty")
        print(f"    {mth:2d}: {hrs:4d} / {exp} = {cap*100:5.1f}%  {mark}")

shared = set(b143) & set(b055)
print(f"\nSimultaneous valid hours: {len(shared)}")

p19 = panel[(panel["city"] == "Paris") & (panel["date"].dt.year == 2019)]
both = p19[p19["station"].isin(["FR04143", "FR04055"])].groupby("date")["station"].nunique()
print("2019 months where both clear the capture rule:", int((both == 2).sum()))
for st in ["FR04143", "FR04055"]:
    s = panel[panel["station"] == st]["date"]
    print(f"  {st} panel span: {s.min().date()} to {s.max().date()}")
```

### 10e. Descriptive offsets across the splice

Because FR04143 and FR04055 do not overlap, their level difference cannot be estimated directly. I compare each one with the continuous traffic station FR04329 as a descriptive common reference. FR04329 does not enter the attribution model.

I report full-segment means and local windows around the handover. They disagree in sign, showing that the inferred offset is sensitive to the period chosen. The local windows also contain fewer than twelve observations because some months fail the capture rule.

As a sensitivity check, I remove March-May 2020 from the incoming window. The estimate becomes more negative, but this does not identify what caused the level difference because the stations never overlap.

The EEA metadata printed in section 4 identify a TEOM-FDMS record for FR04143. FR04055 has records for both TEOM-FDMS and FIDAS optical sampling points. The Geod'Air station export used here does not identify which sampling point produced each reported value, so a change in measurement principle cannot be ruled out. Siting and a genuine local concentration difference are also possible. The stations do not overlap, so these calculations describe the uncertainty rather than repairing or identifying the splice.

```python
mp = (paris.assign(ym=paris["datetime"].dt.to_period("M"))
        .groupby(["code site", "ym"])["valeur brute"].agg(["mean","count"]).reset_index())
mp["expected"] = mp["ym"].apply(lambda p: p.days_in_month * 24)
mp = mp[mp["count"] / mp["expected"] >= 0.75]
piv = mp.pivot(index="ym", columns="code site", values="mean")

d143 = (piv["FR04143"] - piv["FR04329"]).dropna()
d055 = (piv["FR04055"] - piv["FR04329"]).dropna()
loc143 = d143[(d143.index >= pd.Period("2018-09")) & (d143.index <= pd.Period("2019-08"))]
loc055 = d055[(d055.index >= pd.Period("2019-12")) & (d055.index <= pd.Period("2020-11"))]

print("Segment means:")
print(f"  FR04143 - FR04329, {d143.index.min()} to {d143.index.max()}: "
      f"{d143.mean():+.3f} (sd {d143.std():.3f}, n={len(d143)})")
print(f"  FR04055 - FR04329, {d055.index.min()} to {d055.index.max()}: "
      f"{d055.mean():+.3f} (sd {d055.std():.3f}, n={len(d055)})")
print(f"  difference: {d055.mean() - d143.mean():+.3f}")

print("\nBoundary-local (twelve months either side):")
print(f"  FR04143 - FR04329, 2018-09 to 2019-08: {loc143.mean():+.3f} "
      f"(sd {loc143.std():.3f}, n={len(loc143)})")
print(f"  FR04055 - FR04329, 2019-12 to 2020-11: {loc055.mean():+.3f} "
      f"(sd {loc055.std():.3f}, n={len(loc055)})")
print(f"  difference: {loc055.mean() - loc143.mean():+.3f}")

alt = loc055.drop([p for p in loc055.index
                   if pd.Period("2020-03") <= p <= pd.Period("2020-05")])
print(f"\n  FR04055 - FR04329 excluding 2020-03 to 2020-05: {alt.mean():+.3f} "
      f"(sd {alt.std():.3f}, n={len(alt)})")
print("  The estimate is more negative after omitting March-May 2020.")

print("\nMonth by month, boundary-local:")
print("  FR04143:", ", ".join(f"{v:+.2f}" for v in loc143.values))
print("  FR04055:", ", ".join(f"{v:+.2f}" for v in loc055.values))
```
